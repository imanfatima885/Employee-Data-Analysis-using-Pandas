# Employee Data Analysis using Python and Pandas

```python
import pandas as pd
```

```python
# --------------------------------------------------
# 1. Import the dataset
# --------------------------------------------------

employee_data = pd.read_csv("employee_data.csv")
```

```python
print("=" * 60)
print("EMPLOYEE DATA ANALYSIS")
print("=" * 60)
```

```python
# --------------------------------------------------
# 2. Display the first 5 records
# --------------------------------------------------

print("\n1. First 5 Records:")
employee_data.head()
```

```python
# --------------------------------------------------
# 3. Display the last 5 records
# --------------------------------------------------

print("\n2. Last 5 Records:")
employee_data.tail()
```

```python
# --------------------------------------------------
# 4. Check dataset information
# --------------------------------------------------

print("\n3. Dataset Information:")
employee_data.info()
```

```python
# --------------------------------------------------
# 5. Display summary statistics
# --------------------------------------------------

print("\n4. Summary Statistics:")
print(employee_data.describe())
```

```python
# --------------------------------------------------
# 6. Calculate the average salary
# --------------------------------------------------

average_salary = employee_data["Salary"].mean()

print("\n5. Average Salary:")
print(f"{average_salary:,.2f}")
```

```python
# --------------------------------------------------
# 7. Find the employee with the highest salary
# --------------------------------------------------

highest_salary_employee = employee_data.loc[
    employee_data["Salary"].idxmax()
]

print("\n6. Employee with the Highest Salary:")
print(highest_salary_employee)
```

```python
# --------------------------------------------------
# 8. Find the employee with the most experience
# --------------------------------------------------

most_experienced_employee = employee_data.loc[
    employee_data["Experience (Years)"].idxmax()
]

print("\n7. Employee with the Most Experience:")
print(most_experienced_employee)
```

```python
# --------------------------------------------------
# 9. Display employees with salary greater than 80,000
# --------------------------------------------------

high_salary_employees = employee_data[
    employee_data["Salary"] > 80000
]

print("\n8. Employees with Salary Greater Than 80,000:")
high_salary_employees
```

```python
# --------------------------------------------------
# 10. Find department with the highest average salary
# --------------------------------------------------

department_average_salary = employee_data.groupby(
    "Department"
)["Salary"].mean()

highest_average_salary_department = department_average_salary.idxmax()
highest_department_salary = department_average_salary.max()

print("\n9. Department with the Highest Average Salary:")
print(f"Department: {highest_average_salary_department}")
print(f"Average Salary: {highest_department_salary:,.2f}")
```

```python
# --------------------------------------------------
# 11. Count employees in each department
# --------------------------------------------------

department_employee_count = employee_data[
    "Department"
].value_counts()

print("\n10. Number of Employees in Each Department:")
print(department_employee_count)
```

```python
# --------------------------------------------------
# 12. Sort dataset by salary in descending order
# --------------------------------------------------

sorted_employee_data = employee_data.sort_values(
    by="Salary",
    ascending=False
)

print("\n11. Employees Sorted by Salary (Descending):")
sorted_employee_data
```

```python
# --------------------------------------------------
# 13. Save the analyzed dataset
# --------------------------------------------------

sorted_employee_data.to_csv(
    "employee_analysis.csv",
    index=False
)

print("\n12. Analysis completed successfully!")
print("The analyzed dataset has been saved as 'employee_analysis.csv'.")
```
